```python
LAST_NAME = "CORALDE"
STUDENT_ID = "TUPM-26-1903"
FAVORITE_ARTIST = "ARIANA GRANDE"

SEED_DIGIT = 3
ID_SUM = 21

NAME_LENGTH = len(LAST_NAME)

ARTIST_LENGTH = len(FAVORITE_ARTIST.replace(" ", ""))

ARTIST_SUM = 0 

for ch in FAVORITE_ARTIST.upper():
    if ch.isalpha():
        ARTIST_SUM += ord(ch) - ord('A') + 1

print("LAST_NAME =", LAST_NAME)
print("STUDENT_ID =", STUDENT_ID)
print("NAME_LENGTH =", NAME_LENGTH)

print("FAVORITE-ARTIST =", FAVORITE_ARTIST)
print("ARTIST_LENGTH =", ARTIST_LENGTH)

print("SEED_DIGIT =", SEED_DIGIT)
print("ID_SUM =", ID_SUM)

print("ARTIST_SUM =", ARTIST_SUM)
```

```python
print("============================")
print("EXERCISE 1")
print("============================")

records = []

for i in range(1, 21):
    records.append(ID_SUM + i * 5)

rotation = SEED_DIGIT

rotated = records[rotation:] + records[:rotation]

target = ID_SUM + ARTIST_SUM

comparison = 0 
position = -1 

for i in range(len(rotated)):
    comparison += 1

    if rotated[i] == target:
        position = i
        break

print("Original Dataset:")
print(records)

print("\nRotated Dataset:")
print(rotated)

print("\nTarget Incident:", target)

if position != -1:
    print("Positon:", position)
else:
    print("Target Not Found")

print("Comparison Count:", comparison)

print("\nTime Complexity: O(n)")
print("Space Complexity: O(1)")
```

```python
print("============================")
print("EXERCISE 2")
print("UNIVERSITY ENROLLMENT RECORDS")
print("============================")

dataset_a = []

for x in range(10):
    dataset_a.append(SEED_DIGIT + ID_SUM + (x * 3))

dataset_b = []

for x in range(10):
    dataset_b.append(NAME_LENGTH + ARTIST_LENGTH + (x * 4))

merged = dataset_a + dataset_b

merged.sort

comparison = len(dataset_a) + len(dataset_b)

print("\nDepartment A Dataset:")
print(dataset_a)

print("\nDepartment B Dataset:")
print(dataset_b)

print("\nConsolidated Dataset:")
print(merged)

print("\nRecords Before Consolidation:")
print(len(dataset_a) + len(dataset_b))

print("\nRecords After Consolidation:")
print(len(merged))

print("\nComparison Count:")
print(comparison)

print("\nOrdering Verification:")
print(merged == sorted(merged))

print("\nTime Complexity: O(n log n)")
print("Space Complexity: in O(n)")

```

```python
import time 

print("============================")
print("EXERCISE 3")
print("E-COMMERCE ORDER PROCESSING")
print("============================")

orders = [29, 41, 17, 55, 33, 72, 11, 64]

print("\nOriginal Orders:")
print(orders)

bubble_data = orders.copy

start = time.perf_counter()
bubble_data = orders.copy()
bubble_time = time.perf_counter() - start

selection_data = orders.copy

start = time.perf_counter()
selection_data = orders.copy()
selection_time = time.perf_counter() - start

insertion_data = orders.copy

start = time.perf_counter()
insertion_data = orders.copy()
insertion_time = time.perf_counter() - start

print("\nBubble Sort Result:")
print(bubble_data)
print("Execution Time:", bubble_time)

print("\nSelection Sort Result:")
print(selection_data)
print("Execution Time:", selection_time)

print("\nInsertion Sort Result:")
print(insertion_data)
print("Execution Time:", insertion_time)

print("\nComplexity Analysis")
print("Bubble Sort: O(n^2)")
print("Selection Sort: O(n^2)")
print("Insertion Sort: O(n^2)")
```

```python
print("============================")
print("EXERCISE 1")
print("============================")

ID_SUM = sum(ord(char) - 64 for char in "CORALDE")
SEED_DIGIT = 3
ARTIST_SUM = sum(ord(char) - 64 for char in "ARIANA GRANDE")

records = []

for i in range(1, 21):
    records.append(ID_SUM + i * 5)

rotation = SEED_DIGIT
rotated = records[rotation:] + records[:rotation]

target = ID_SUM + ARTIST_SUM

comparison = 0
position = -1

for i in range(len(rotated)):
    comparison += 1

    if rotated[i] == target:
        position = i
        break

print("\nOriginal Dataset:")
print(records)

print("\nRotated Dataset:")
print(rotated)

print("\nTarget Incident:")
print(target)

print("\nComparison Counts:", comparison)
print("Linear Search Comparison Count:", comparison)

# Time and Space Complexity
print("\nTime and Space Complexity:")
print("Time Complexity: O(n)")
print("Space Complexity: O(n)")

# Final Output
print("\n============================")
print("FINAL OUTPUT")
print("============================")

if position == -1:
    print("Target Found at Position:", position)
else:
    print("Target Not Found")

print("Total Comparisons:", comparison)
print("============================")
```

```python
import time
import random

print("============================")
print("EXERCISE 3")
print("E-COMMERCE ORDER PROCESSING")
print("============================")

orders = [29, 41, 17, 55, 33, 72, 11, 64]

print("\nGenerated Orders:")
print(orders)

conditions = {
    "Sorted": sorted(orders),
    "Reverse-Ordered": sorted(orders, reverse=True),
    "Random": random.Random(3).sample(orders, len(orders))
}
def bubble_sort(data):
    arr = data.copy()
    comparisons = movements = 0

    for i in range(len(arr) - 1):
        for j in range(len(arr) - 1 - i):
            comparisons += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 2

    return arr, comparisons, movements

def selection_sort(data):
    arr = data.copy()
    comparisons = movements = 0

    for i in range(len(arr) - 1):
        minimum = i
        for j in range(i + 1, len(arr)):
            comparisons += 1
            if arr[j] < arr[minimum]:
                minimum = j

        if minimum != i:
            arr[i], arr[minimum] = arr[minimum], arr[i]
            movements += 2

    return arr, comparisons, movements

def insertion_sort(data):
    arr = data.copy()
    comparisons = movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        if j + 1 != i:
            arr[j + 1] = key
            movements += 1

    return arr, comparisons, movements

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

# Process and Display Results
for condition, data in conditions.items():
    print("\n----------------------------")
    print(condition, "Input:")
    print(data)

    for name, algorithm in algorithms.items():
        start = time.perf_counter()
        result, comparisons, movements = algorithm(data)
        elapsed = time.perf_counter() - start

        print("\n", name)
        print("Result:", result)
        print("Comparisons:", comparisons)
        print("Movements:", movements)
        print("Execution Time:", elapsed)
        print("Sorting Verification:", result == sorted(data))

print("\n============================")
print("BIG O ANALYSIS")
print("============================")
print("Bubble Sort: Best O(n), Average O(n^2), Worst O(n^2)")
print("Selection Sort: Best O(n^2), Average O(n^2), Worst O(n^2)")
print("Insertion Sort: Best O(n), Average O(n^2), Worst O(n^2)")

print("\n============================")
print("CASE OBSERVATIONS")
print("============================")
print("Best Case: Sorted input helps Bubble and Insertion Sort.")
print("Average Case: Random input requires varying comparisons and movements.")
print("Worst Case: Reverse-ordered input is difficult for Bubble and Insertion Sort.")

print("\n============================")
print("FINAL OUTPUT")
print("============================")
print("All three algorithms were tested on all three input conditions.")
print("Comparisons, movements, and execution times were recorded.")
print("Sorting verification and Big O analysis were completed.")
```
